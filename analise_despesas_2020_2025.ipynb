### Comportamento e pontos de atenção
- O total líquido analisado foi **R$ 1.333.828.027,97**. O gasto anual passou de R$ 172,9 milhões em 2020 para R$ 249,4 milhões em 2024; em 2025 caiu 4,0% frente a 2024, mas permaneceu 38,4% acima de 2020. A comparação é nominal, sem correção pela inflação.
- Nordeste e Sudeste concentraram os maiores totais anuais. Em 2025, corresponderam a aproximadamente 30,9% e 32,3% do gasto incluído, respectivamente. Centro-Oeste e Sul tiveram maior crescimento acumulado no período, cerca de 59% e 58%.
- Há lançamentos individuais altos na categoria de divulgação da atividade parlamentar. O maior observado foi R$ 184.428,00 (Duda Ramos, RR, 2024), contra limite estatístico de triagem de R$ 27.880,00 para aquela categoria/ano. Valores assim pedem conferência do documento, fornecedor e período coberto; não são, isoladamente, evidência de irregularidade.
- A regra de triagem é `vlrLiquido > Q3 + 3 x IQR`, calculada separadamente por categoria e ano. Ela gera milhares de candidatos por ano, portanto serve para priorizar revisão e precisa de validação documental.

### Escopo
Foram usados os CSVs compactados de 2020 a 2025, que têm 12 meses registrados em cada ano. O gasto é a soma de `vlrLiquido` após glosas; restituições e glosas são apresentadas à parte na auditoria e não são subtraídas novamente. Linhas sem nome de parlamentar ou sem UF brasileira válida foram excluídas dos rankings e totais regionais.

### Conclusões sobre comportamento e tendências

- **Evolução do gasto total:** os valores nominais cresceram 21,0% em 2021, 4,5% em 2022, 11,9% em 2023 e 2,0% em 2024; em 2025 houve queda de 4,0%. Portanto, a redução mais recente ocorre depois de quatro anos de alta, e não apaga o crescimento acumulado de 38,4% desde 2020.
- **Ritmo regional:** entre 2020 e 2025, Centro-Oeste teve o maior aumento relativo (+59,8%), seguido do Sul (+58,5%). Ainda assim, em 2025 essas regiões representaram 8,0% e 15,3% do total incluído, enquanto Sudeste e Nordeste somaram 32,3% e 30,9%. Crescimento percentual e participação no gasto são medidas diferentes.
- **Lideranças anuais:** o líder do Norte mudou nos seis anos; no Nordeste, General Girão liderou em 2023 e 2025; no Centro-Oeste, Beto Pereira liderou em 2021 e 2023; no Sudeste, Gilberto Nascimento liderou em 2022 e 2024; no Sul, Pompeo de Mattos liderou de 2023 a 2025.
- **Concentração por UF no top 4 anual:** MS apareceu em 18 de 24 posições do Centro-Oeste e RS em 21 de 24 posições do Sul. No Sudeste, SP ocupou 18 de 24 posições; no Norte, RR ocupou 13 de 24. No Nordeste, RN apareceu 9 vezes, distribuídas pelos seis anos.

As contagens de UF acima se referem somente às 24 posições do top 4 em cada região ao longo dos seis anos. Elas não representam a participação da UF no gasto regional total, nem demonstram irregularidade.

### Validação e limitações

Foi feito backtest de origem móvel para 2024 e 2025. Em 2024, o Ridge teve WAPE de **5,12%**, contra **4,38%** do baseline que repete o gasto do ano anterior. Em 2025, o Ridge teve **5,28%**, contra **6,11%** do baseline. No conjunto dos dois anos, os erros foram praticamente iguais: **5,20% para Ridge e 5,22% para o baseline**.

As estimativas são um cenário, não um orçamento garantido: há somente seis observações anuais por parlamentar e nenhum ajuste inflacionário. Os subtotais representam apenas a coorte de vinte maiores acumulados, não o gasto total das regiões nem de toda a Câmara.

```python
from collections import defaultdict
from pathlib import Path
from zipfile import ZipFile
import csv
import io
import re

PASTA_DADOS = Path(r'C:\Users\Vitor\Downloads\projeto de análise dos parlamentares para 2026 e 2027\dados_camara')
ANOS = range(2020, 2026)
REGIOES = {
    'Norte': {'AC', 'AP', 'AM', 'PA', 'RO', 'RR', 'TO'},
    'Nordeste': {'AL', 'BA', 'CE', 'MA', 'PB', 'PE', 'PI', 'RN', 'SE'},
    'Centro-Oeste': {'DF', 'GO', 'MT', 'MS'},
    'Sudeste': {'ES', 'MG', 'RJ', 'SP'},
    'Sul': {'PR', 'RS', 'SC'},
}
UF_REGIAO = {uf: regiao for regiao, ufs in REGIOES.items() for uf in ufs}
COLUNAS = [
    'txNomeParlamentar', 'nuDeputadoId', 'sgUF', 'sgPartido', 'vlrLiquido',
    'vlrGlosa', 'vlrRestituicao', 'numMes', 'txtDescricao', 'txtFornecedor',
]


def limpar_texto(valor):
    return re.sub(r'[\x00-\x1f\x7f-\x9f\ufffd]', '', valor or '').strip()


def numero(valor):
    try:
        return float((valor or '').strip().replace(',', '.'))
    except ValueError:
        return 0.0


def ler_lancamentos(ano):
    caminho = PASTA_DADOS / f'Ano-{ano}.csv.zip'
    with ZipFile(caminho) as pacote:
        nome_csv = next(nome for nome in pacote.namelist() if nome.lower().endswith('.csv'))
        with pacote.open(nome_csv) as binario:
            binario.read(3)
            texto = io.TextIOWrapper(binario, encoding='utf-8', errors='replace', newline='')
            leitor = csv.reader(texto, delimiter=';')
            cabecalho = next(leitor)
            indices = {coluna: cabecalho.index(coluna) for coluna in COLUNAS}
            for linha in leitor:
                yield {coluna: linha[indice] if indice < len(linha) else '' for coluna, indice in indices.items()}


def quantil(valores, proporcao):
    ordenados = sorted(valores)
    posicao = (len(ordenados) - 1) * proporcao
    inferior = int(posicao)
    superior = min(inferior + 1, len(ordenados) - 1)
    fracao = posicao - inferior
    return ordenados[inferior] * (1 - fracao) + ordenados[superior] * fracao


totais_parlamentar = defaultdict(float)
totais_parlamentar_ano = defaultdict(float)
anos_ativos = defaultdict(set)
metadados_parlamentar = {}
totais_regiao_ano = defaultdict(float)
totais_mes_ano = defaultdict(float)
anomalias = []
qualidade = []

for ano in ANOS:
    caminho = PASTA_DADOS / f'Ano-{ano}.csv.zip'
    if not caminho.exists():
        raise FileNotFoundError(caminho)
    registros_categoria = defaultdict(list)
    linhas_arquivo = 0
    linhas_validas = 0
    negativos = 0
    total_glosas = 0.0
    total_restituicoes = 0.0
    meses = set()

    for lancamento in ler_lancamentos(ano):
        linhas_arquivo += 1
        uf = limpar_texto(lancamento['sgUF']).upper()
        parlamentar = limpar_texto(lancamento['txNomeParlamentar'])
        if uf not in UF_REGIAO or not parlamentar:
            continue
        valor = numero(lancamento['vlrLiquido'])
        regiao = UF_REGIAO[uf]
        id_deputado = lancamento['nuDeputadoId'].strip().removesuffix('.0')
        chave = (id_deputado or parlamentar.casefold(), uf)
        categoria = limpar_texto(lancamento['txtDescricao']) or 'Sem categoria'
        partido = limpar_texto(lancamento['sgPartido'])
        mes = int(numero(lancamento['numMes']))

        totais_parlamentar[chave] += valor
        totais_parlamentar_ano[(chave, ano)] += valor
        anos_ativos[chave].add(ano)
        metadados_parlamentar[chave] = {
            'parlamentar': parlamentar, 'partido': partido, 'regiao': regiao,
        }
        totais_regiao_ano[(regiao, ano)] += valor
        if 1 <= mes <= 12:
            totais_mes_ano[(ano, mes)] += valor
            meses.add(mes)
        registros_categoria[categoria].append((
            valor, parlamentar, uf, limpar_texto(lancamento['txtFornecedor']),
        ))
        linhas_validas += 1
        negativos += valor < 0
        total_glosas += numero(lancamento['vlrGlosa'])
        total_restituicoes += numero(lancamento['vlrRestituicao'])

    candidatos_ano = []
    for categoria, registros in registros_categoria.items():
        valores = [registro[0] for registro in registros]
        q1 = quantil(valores, 0.25)
        q3 = quantil(valores, 0.75)
        limite = q3 + 3 * (q3 - q1)
        for valor, parlamentar, uf, fornecedor in registros:
            if valor > limite:
                candidatos_ano.append({
                    'ano': ano, 'parlamentar': parlamentar, 'UF': uf,
                    'categoria': categoria, 'fornecedor': fornecedor,
                    'valor': valor, 'limite': limite,
                })
    anomalias.extend(sorted(candidatos_ano, key=lambda item: item['valor'], reverse=True)[:20])
    qualidade.append({
        'ano': ano, 'linhas_arquivo': linhas_arquivo, 'linhas_validas': linhas_validas,
        'fora_escopo': linhas_arquivo - linhas_validas, 'lancamentos_negativos': negativos,
        'glosas': total_glosas, 'restituicoes': total_restituicoes,
        'meses': sorted(meses), 'candidatos_outlier': len(candidatos_ano),
    })
    print(f'{ano}: {linhas_arquivo:,} linhas; {linhas_validas:,} com parlamentar e UF válidos; {len(candidatos_ano):,} candidatos a outlier')

print(f'\\nTotal líquido analisado: R$ {sum(totais_regiao_ano.values()):,.2f}')
```

```python
def moeda(valor):
    texto = f'{valor:,.2f}'
    return 'R$ ' + texto.replace(',', 'X').replace('.', ',').replace('X', '.')

ordem_regioes = ['Norte', 'Nordeste', 'Centro-Oeste', 'Sudeste', 'Sul']
ranking_anual = []
for ano in ANOS:
    for regiao in ordem_regioes:
        chaves = [
            chave for chave, dados in metadados_parlamentar.items()
            if dados['regiao'] == regiao and (chave, ano) in totais_parlamentar_ano
        ]
        chaves.sort(
            key=lambda chave: totais_parlamentar_ano[(chave, ano)],
            reverse=True,
        )
        for posicao, chave in enumerate(chaves[:4], start=1):
            dados = metadados_parlamentar[chave]
            ranking_anual.append((
                ano, regiao, posicao, dados['parlamentar'], chave[1],
                totais_parlamentar_ano[(chave, ano)],
            ))

print('TOP 4 POR REGIÃO EM CADA ANO (vlrLiquido anual)')
print('Ano | Região | 1º | 2º | 3º | 4º')
for ano in ANOS:
    for regiao in ordem_regioes:
        linhas = [linha for linha in ranking_anual if linha[0] == ano and linha[1] == regiao]
        colocados = [
            f'{nome} ({uf}): {moeda(gasto)}'
            for _, _, _, nome, uf, gasto in linhas
        ]
        print(f'{ano} | {regiao} | ' + ' | '.join(colocados))

print('\\nGASTO LÍQUIDO POR REGIÃO E ANO')
print('Região | ' + ' | '.join(str(ano) for ano in ANOS))
for regiao in ordem_regioes:
    valores = [totais_regiao_ano[(regiao, ano)] for ano in ANOS]
    print(regiao + ' | ' + ' | '.join(moeda(valor) for valor in valores))

print('\\nTOTAL ANUAL E VARIAÇÃO NOMINAL')
totais_anuais = [sum(totais_regiao_ano[(regiao, ano)] for regiao in ordem_regioes) for ano in ANOS]
for indice, ano in enumerate(ANOS):
    variacao = '—' if indice == 0 else f'{(totais_anuais[indice] / totais_anuais[indice - 1] - 1) * 100:+.1f}%'
    print(f'{ano}: {moeda(totais_anuais[indice])} ({variacao})')
print(f"Mudança acumulada 2020–2025: {(totais_anuais[-1] / totais_anuais[0] - 1) * 100:+.1f}% (nominal)")

print('\\nMÊS DE MAIOR GASTO EM CADA ANO')
for ano in ANOS:
    mes, gasto = max(
        ((mes, valor) for (ano_lancamento, mes), valor in totais_mes_ano.items() if ano_lancamento == ano),
        key=lambda item: item[1],
    )
    print(f'{ano}: mês {mes}, {moeda(gasto)}')
```

```python
# Projeção individual dos 20 parlamentares do top 4 de cada região.
import math

parlamentares_modelo = []
for regiao in ordem_regioes:
    chaves_regiao = [
        chave for chave, dados in metadados_parlamentar.items()
        if dados['regiao'] == regiao
    ]
    chaves_regiao.sort(key=lambda chave: totais_parlamentar[chave], reverse=True)
    for chave in chaves_regiao[:4]:
        parlamentares_modelo.append({
            'chave': chave,
            'regiao': regiao,
            'parlamentar': metadados_parlamentar[chave]['parlamentar'],
            'UF': chave[1],
            'historico': [totais_parlamentar_ano.get((chave, ano), 0.0) / 1000 for ano in ANOS],
        })


def atributos_parlamentar(indice, ano_indice):
    individual = [float(posicao == indice) for posicao in range(len(parlamentares_modelo))]
    return [*individual, ano_indice]


def resolver_sistema(matriz, vetor):
    aumentado = [linha[:] + [alvo] for linha, alvo in zip(matriz, vetor)]
    tamanho = len(vetor)
    for coluna in range(tamanho):
        pivo = max(range(coluna, tamanho), key=lambda linha: abs(aumentado[linha][coluna]))
        aumentado[coluna], aumentado[pivo] = aumentado[pivo], aumentado[coluna]
        divisor = aumentado[coluna][coluna]
        aumentado[coluna] = [valor / divisor for valor in aumentado[coluna]]
        for linha in range(tamanho):
            if linha == coluna:
                continue
            fator = aumentado[linha][coluna]
            aumentado[linha] = [
                atual - fator * pivo_valor
                for atual, pivo_valor in zip(aumentado[linha], aumentado[coluna])
            ]
    return [aumentado[linha][-1] for linha in range(tamanho)]


def ajustar_modelo_ridge(atributos, alvos, alpha=100):
    quantidade_atributos = len(atributos[0])
    medias = [sum(linha[coluna] for linha in atributos) / len(atributos) for coluna in range(quantidade_atributos)]
    desvios = [
        math.sqrt(sum((linha[coluna] - medias[coluna]) ** 2 for linha in atributos) / len(atributos)) or 1.0
        for coluna in range(quantidade_atributos)
    ]
    padronizados = [
        [(linha[coluna] - medias[coluna]) / desvios[coluna] for coluna in range(quantidade_atributos)]
        for linha in atributos
    ]
    media_alvo = sum(alvos) / len(alvos)
    matriz = [
        [
            sum(linha[i] * linha[j] for linha in padronizados) + (alpha if i == j else 0.0)
            for j in range(quantidade_atributos)
        ]
        for i in range(quantidade_atributos)
    ]
    vetor = [
        sum(linha[coluna] * (alvo - media_alvo) for linha, alvo in zip(padronizados, alvos))
        for coluna in range(quantidade_atributos)
    ]
    return medias, desvios, media_alvo, resolver_sistema(matriz, vetor)


def prever_modelo_ridge(modelo, atributos):
    medias, desvios, media_alvo, coeficientes = modelo
    padronizados = [
        (valor - medias[indice]) / desvios[indice]
        for indice, valor in enumerate(atributos)
    ]
    return media_alvo + sum(
        valor * coeficiente for valor, coeficiente in zip(padronizados, coeficientes)
    )


def treinar_prever_no_corte(indice_teste):
    atributos = []
    alvos = []
    for indice, parlamentar in enumerate(parlamentares_modelo):
        for indice_ano in range(indice_teste):
            atributos.append(atributos_parlamentar(indice, indice_ano))
            alvos.append(parlamentar['historico'][indice_ano])
    modelo = ajustar_modelo_ridge(atributos, alvos)
    previsoes = [
        prever_modelo_ridge(modelo, atributos_parlamentar(indice, indice_teste))
        for indice in range(len(parlamentares_modelo))
    ]
    return modelo, previsoes


# Dois backtests de origem móvel: 2024 e 2025.
erros_ridge = []
erros_baseline = []
valores_observados = []
for indice_teste in (4, 5):
    _, previsoes_teste = treinar_prever_no_corte(indice_teste)
    erros_fold = []
    erros_baseline_fold = []
    observados_fold = []
    for indice, parlamentar in enumerate(parlamentares_modelo):
        observado = parlamentar['historico'][indice_teste]
        erros_fold.append(abs(observado - previsoes_teste[indice]))
        erros_baseline_fold.append(abs(observado - parlamentar['historico'][indice_teste - 1]))
        observados_fold.append(observado)
    erros_ridge.extend(erros_fold)
    erros_baseline.extend(erros_baseline_fold)
    valores_observados.extend(observados_fold)
    wape_fold = sum(erros_fold) / sum(observados_fold) * 100
    wape_baseline_fold = sum(erros_baseline_fold) / sum(observados_fold) * 100
    print(f'Backtest {2020 + indice_teste}: WAPE Ridge {wape_fold:.2f}% | baseline {wape_baseline_fold:.2f}%')

wape_ridge = sum(erros_ridge) / sum(valores_observados) * 100
wape_baseline = sum(erros_baseline) / sum(valores_observados) * 100
print(f'Backtest combinado 2024–2025: Ridge {wape_ridge:.2f}% | baseline {wape_baseline:.2f}%')

modelo_final, _ = treinar_prever_no_corte(6)
for indice, parlamentar in enumerate(parlamentares_modelo):
    parlamentar['previsao_2026'] = max(
        0.0, prever_modelo_ridge(modelo_final, atributos_parlamentar(indice, 6)) * 1000,
    )
    parlamentar['previsao_2027'] = max(
        0.0, prever_modelo_ridge(modelo_final, atributos_parlamentar(indice, 7)) * 1000,
    )

print('\\nPREVISÃO DE GASTO INDIVIDUAL (R$)')
print('Região | Parlamentar | UF | Gasto 2025 | Previsão 2026 | Previsão 2027')
for parlamentar in parlamentares_modelo:
    print(
        f"{parlamentar['regiao']} | {parlamentar['parlamentar']} | {parlamentar['UF']} | "
        f"{moeda(parlamentar['historico'][5] * 1000)} | "
        f"{moeda(parlamentar['previsao_2026'])} | {moeda(parlamentar['previsao_2027'])}"
    )
print('\\nSUBTOTAL PREVISTO APENAS PARA O TOP 4 DE CADA REGIÃO')
for regiao in ordem_regioes:
    grupo = [p for p in parlamentares_modelo if p['regiao'] == regiao]
    total_2026 = sum(p['previsao_2026'] for p in grupo)
    total_2027 = sum(p['previsao_2027'] for p in grupo)
    print(f'{regiao} | {moeda(total_2026)} | {moeda(total_2027)}')
print('Total top 20 | {} | {}'.format(
    moeda(sum(p['previsao_2026'] for p in parlamentares_modelo)),
    moeda(sum(p['previsao_2027'] for p in parlamentares_modelo)),
))
```
